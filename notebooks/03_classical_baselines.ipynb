# Classical machine-learning baselines

This notebook runs Logistic Regression, RBF-SVM, and XGBoost on the same predefined midterm train/test split. It does not tune hyperparameters using the test set. PR-AUC (average precision) is the primary metric.

## Evaluation protocol

The shared split is the reproducible 80/20 stratified random baseline with seed 42. `StandardScaler` is fitted only on training features. Each model uses its native prediction rule; test labels are not used to choose a threshold. The fixed RBF-SVM iteration cap is recorded with its fit status and warnings.
