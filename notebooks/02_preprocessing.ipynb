# Leakage-free preprocessing and baseline split

This notebook runs reusable preprocessing code for the midterm's conventional stratified random evaluation split. It does not train a model, modify the raw CSV, remove rows, perform feature selection, or use test labels during preprocessing.

## Protocol

- Target: `Class`.
- Baseline features: every non-target column (`Time`, `V1`–`V28`, and `Amount`).
- `Amount`, `Time`, and the anonymized numerical features are standardized with `StandardScaler` fitted on `X_train` only.
- The baseline split is 80/20, stratified by `Class`, shuffled with fixed random seed 42.
- The original unscaled `Time` series is retained with each partition. It will support a separate chronological, non-shuffled protocol later; this notebook does not implement that future experiment.
